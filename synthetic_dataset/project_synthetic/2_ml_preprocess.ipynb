# 2. Preprocess the calibration set

A short, common chain for soil spectra, each step reading the output of the one before:

| Step | Process | Output dataframe |
|---|---|---|
| 1 | `detect_outliers` (isolation forest on c-org, n-tot, ec; values set to missing) | `synthetic_calibration_10nm.olif` |
| 2 | `spectra_scatter_correction` (SNV) | `synthetic_calibration_10nm.olif.snv` |
| 3 | `spectra_scaling` (meancentring) | `synthetic_calibration_10nm.olif.snv.mc` |

Only the calibration set is preprocessed. The model refits steps 2-3 on its training samples and
stores them, so the validation set stays unprocessed.
