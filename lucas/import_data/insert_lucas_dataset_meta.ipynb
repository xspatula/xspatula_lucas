## Insert LUCAS dataset meta data

This notebook is in a customized project for inserting specific meta data for the overall dataset of the LUCAS soil sampling campaigns. The project requires a customized xspatula python module that is not included in the xspatula core. By default the path to the start module for a custom project is ./src/"project_name"; it needs to contain a startup function. The default name of that function is "Run_process". 

This notebook inserts data in the database in a **single-step**. Each cell below translates data straight from excel/csv into the database in one step, using the `insert_tabular_data` process. This happaens in 2 steps; first the excel data is translated to JSON command files and then the JSON command files are read and compared with the database records that tell where each data point should be stored in the database. The intermediate JSON files are stored under sub-directories called `staging` (e.g. ./lucas/import_data/utility/general/insert_process/staging). There is no manual hand-off between steps, and this route is **INSERT-only**: existing records are left untouched, never updated. If you need to UPDATE or DELETE data, or want to inspect/edit the generated JSON before it is applied, you need to use the process `manage_tabular_data`. You can then hand edit the JSON command files under `staging` and point the `manage_tabular_data` process to the particular JSON process file holding the data you want to update or delete. There is also a **two-step** process (see [manual](https://xspatula.github.io)) if you want to edit or inspect the JSON files before inserting the data.
